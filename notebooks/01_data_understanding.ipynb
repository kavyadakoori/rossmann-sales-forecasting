### Missing/Null Value Analysis

```python
print('='*60)
print('Null Values in Train Data: ')
print('='*60)
train_null = pd.DataFrame({'Count of Missing values': train.isnull().sum(),
              '% of Missing Values': (train.isnull().sum()*100/len(train)).round(2)})
display(train_null) if train_null['Count of Missing values'].sum()>0 else print("No Null values in Train Data!")
```

```python
print('='*60)
print('Null Values in Store Data: ')
print('='*60)
store_null = pd.DataFrame({'Count of Missing values': store.isnull().sum(),
              '% of Missing Values': (store.isnull().sum()*100/len(store)).round(2)})
display(store_null[store_null['Count of Missing values']>0]) if store_null['Count of Missing values'].sum()>0 else print("No Null values in Train Data!")
```

```python
print('='*60)
print('Null Values in Test Data: ')
print('='*60)
test_null = pd.DataFrame({'Count of Missing values': test.isnull().sum(),
              '% of Missing Values': (test.isnull().sum()*100/len(test)).round(2)})
display(test_null[test_null['Count of Missing values']>0]) if test_null['Count of Missing values'].sum()>0 else print("No Null values in Train Data!")
```

```python
# CompetitionDistance — fill with median
store['CompetitionDistance'] = store['CompetitionDistance'].fillna(store['CompetitionDistance'].median())

# Competition open since — fill with 0(either no known competitor or details unavailable)
store['CompetitionOpenSinceMonth'] = store['CompetitionOpenSinceMonth'].fillna(0)
store['CompetitionOpenSinceYear']  = store['CompetitionOpenSinceYear'].fillna(0)

# Promo2 fields — fill with 0(No Promo happened)
store['Promo2SinceWeek'] = store['Promo2SinceWeek'].fillna(0)
store['Promo2SinceYear'] = store['Promo2SinceYear'].fillna(0)

# PromoInterval — fill with 'None'
store['PromoInterval'] = store['PromoInterval'].fillna('None')
```

```python
# Impute Open column with 1, keeping it open to predict its sales
test['Open'] = test['Open'].fillna(1)
```

```python
train['Date'] = pd.to_datetime(train['Date'])
test['Date'] = pd.to_datetime(test['Date'])
print ("Train Data Date range: {} to {}".format(train['Date'].min(), train['Date'].max()))
print ("Test Data Date range: {} to {}".format(test['Date'].min(), test['Date'].max()))
```

### Duplicate Values

```python
print("Duplicate Rows in Train :", train.duplicated().sum())
print("Duplicate Rows in Store :", store.duplicated().sum())
print("Duplicate Rows in Test :", test.duplicated().sum())
```

### Merge Train & Store data

```python
df_train = pd.merge(train,store,on = 'Store',how = 'left')
print("📦 train shape        :", train.shape)
print("📦 test shape         :", test.shape)
print("📦 store shape        :", store.shape)
print("📦 merged train shape :", df_train.shape)

print()
print('Train Data')
print("📅 Date range         :", df_train['Date'].min(), "→", df_train['Date'].max())
print("🏪 Total stores       :", df_train['Store'].nunique())
print("📋 Total columns      :", df_train.columns.tolist())
print()

print()
print('Test Data')
print("📅 Date range         :", test['Date'].min(), "→", test['Date'].max())
print("🏪 Total stores       :", test['Store'].nunique())
print("📋 Total columns      :", test.columns.tolist())
print()
print("✅ Merge successful!")
```
